```python
df.columns = (df.columns.str.strip().str.lower().str.replace(" ","_"))
df.rename(columns={
    "price" : "unit_price",
    "invoicedate" : "invoice_date",
    "stockcode" : "stock_code"
}, inplace=True)
```

```python
df = df.dropna(subset=["customer_id"])
```

```python
df["invoice_date"] = pd.to_datetime(df["invoice_date"], errors="coerce")

(df["customer_id"].astype("Int64"))
```

```python
df["invoice_date"].info()
```

```python
df["is_cancellation"] = (df["invoice"].astype(str).str.startswith("c"))
```

```python
df[df["quantity"] < 0]
```

```python
purchases = df[(~df["is_cancellation"]) & (df["quantity"] > 0) & (df["unit_price"] > 0)].copy()
```

```python
purchases
```

```python
purchases["revenue"] = (purchases["quantity"] * purchases["unit_price"])
```

```python
purchases["invoice_date"].info()
```

```python
purchases.to_csv("../data/cleaned_customer_data.csv", index=False)
```
